```python
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
# =============================================================================
# PARÂMETROS CONFIGURÁVEIS - BSD FORMULA
# =============================================================================

# Pesos dos critérios (devem somar 100%)
WEIGHTS = {
    'payout_ratio': 0.30,      # 30%
    'interest_coverage': 0.10,  # 10%
    'cash_flow_ratio': 0.05,    # 5%
    'roe': 0.10,                # 10%
    'div_growth_5y': 0.10,      # 10%
    'earnings_growth_5y': 0.10, # 10%
    'revenue_growth_5y': 0.05,  # 5%
    'dividend_years': 0.05,     # 5%
    'price_strength': 0.05,     # 5%
    'beta': 0.10                # 10%
}

# Parâmetros ideais para pontuação (0-100 pontos por critério)
SCORING_PARAMS = {
    'payout_ratio': {
        'ideal_min': 20,    # Payout muito baixo pode indicar que empresa não valoriza dividendos
        'ideal_max': 60,    # Payout acima de 60% é arriscado
        'critical_max': 100 # Acima de 100% é insustentável
    },
    'interest_coverage': {
        'min_acceptable': 3,  # Mínimo 3x de cobertura
        'ideal': 10           # Ideal é 10x ou mais
    },
    'cash_flow_ratio': {
        'min_acceptable': 0.8,  # Fluxo de caixa deve ser pelo menos 80% do lucro líquido
        'ideal': 1.2            # Ideal é 120% ou mais
    },
    'roe': {
        'min_acceptable': 10,  # ROE mínimo de 10%
        'ideal': 20            # ROE ideal de 20% ou mais
    },
    'div_growth_5y': {
        'min_acceptable': 0,   # Crescimento mínimo aceitável (0% = mantém)
        'ideal': 10            # Crescimento ideal de 10% ao ano
    },
    'earnings_growth_5y': {
        'min_acceptable': 0,
        'ideal': 15
    },
    'revenue_growth_5y': {
        'min_acceptable': 0,
        'ideal': 10
    },
    'dividend_years': {
        'min_acceptable': 5,   # Mínimo 5 anos pagando dividendos
        'ideal': 25            # Ideal é 25+ anos (Dividend Aristocrat)
    },
    'price_strength': {
        'min_acceptable': -10,  # Retorno mínimo vs mercado (%)
        'ideal': 10             # Ideal é superar mercado em 10%
    },
    'beta': {
        'ideal': 0.8,          # Beta ideal (baixa volatilidade)
        'max_acceptable': 1.5   # Beta máximo aceitável
    }
}
```

```python
# =============================================================================
# FUNÇÕES AUXILIARES
# =============================================================================

def calculate_score(value, criterion):
    """Calcula pontuação de 0-100 para um critério específico"""
    if pd.isna(value) or value is None:
        return 0
    
    params = SCORING_PARAMS[criterion]
    
    if criterion == 'payout_ratio':
        if value < 0:
            return 0
        elif value <= params['ideal_min']:
            if params['ideal_min'] == 0:
                return 100
            return 50 + (value / params['ideal_min']) * 50
        elif value <= params['ideal_max']:
            return 100
        elif value <= params['critical_max']:
            return 100 - ((value - params['ideal_max']) / (params['critical_max'] - params['ideal_max'])) * 100
        else:
            return 0
    
    elif criterion in ['interest_coverage', 'cash_flow_ratio', 'roe', 
                       'div_growth_5y', 'earnings_growth_5y', 'revenue_growth_5y', 
                       'dividend_years', 'price_strength']:
        min_acc = params['min_acceptable']
        ideal = params['ideal']
        
        # Proteção contra divisão por zero
        if min_acc == 0:
            if value >= ideal:
                return 100
            else:
                return max(0, min(100, (value / ideal) * 100)) if ideal != 0 else 0
        
        if value < min_acc:
            return max(0, 50 * (value / min_acc))
        elif value >= ideal:
            return 100
        else:
            # Proteção adicional
            denominator = ideal - min_acc
            if denominator == 0:
                return 100
            return 50 + 50 * ((value - min_acc) / denominator)
    
    elif criterion == 'beta':
        if value <= params['ideal']:
            return 100
        elif value <= params['max_acceptable']:
            denominator = params['max_acceptable'] - params['ideal']
            if denominator == 0:
                return 100
            return 100 - ((value - params['ideal']) / denominator) * 50
        else:
            return max(0, 50 - (value - params['max_acceptable']) * 10)
    
    return 0

def get_stock_data(ticker):
    """Obtém dados fundamentais de uma ação"""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        
        # Dados financeiros
        financials = stock.financials
        balance_sheet = stock.balance_sheet
        cash_flow = stock.cashflow
        
        # Histórico de dividendos
        dividends = stock.dividends
        
        # Preço histórico
        hist = stock.history(period="5y")
        
        return stock, info, financials, balance_sheet, cash_flow, dividends, hist
    except Exception as e:
        print(f"Erro ao obter dados de {ticker}: {str(e)}")
        return None, None, None, None, None, None, None

def calculate_metrics(ticker):
    """Calcula todos os 10 critérios BSD para uma ação"""
    stock, info, financials, balance_sheet, cash_flow, dividends, hist = get_stock_data(ticker)
    
    if info is None:
        return None
    
    metrics = {
        'ticker': ticker,
        'name': info.get('longName', ticker)
    }
    
    # 1. Payout Ratio (30%)
    payout = info.get('payoutRatio', np.nan)
    metrics['payout_ratio'] = payout * 100 if not pd.isna(payout) else np.nan
    
    # 2. Interest Coverage (10%)
    try:
        if not financials.empty and not financials.columns.empty:
            ebit = financials.loc['EBIT', financials.columns[0]] if 'EBIT' in financials.index else np.nan
            interest = financials.loc['Interest Expense', financials.columns[0]] if 'Interest Expense' in financials.index else np.nan
            if not pd.isna(ebit) and not pd.isna(interest) and interest != 0:
                metrics['interest_coverage'] = abs(ebit / interest)
            else:
                metrics['interest_coverage'] = np.nan
        else:
            metrics['interest_coverage'] = np.nan
    except:
        metrics['interest_coverage'] = np.nan
    
    # 3. Cash Flow Ratio (5%)
    try:
        if not cash_flow.empty and not financials.empty:
            operating_cf = cash_flow.loc['Operating Cash Flow', cash_flow.columns[0]] if 'Operating Cash Flow' in cash_flow.index else np.nan
            net_income = financials.loc['Net Income', financials.columns[0]] if 'Net Income' in financials.index else np.nan
            if not pd.isna(operating_cf) and not pd.isna(net_income) and net_income != 0:
                metrics['cash_flow_ratio'] = operating_cf / net_income
            else:
                metrics['cash_flow_ratio'] = np.nan
        else:
            metrics['cash_flow_ratio'] = np.nan
    except:
        metrics['cash_flow_ratio'] = np.nan
    
    # 4. ROE (10%)
    roe = info.get('returnOnEquity', np.nan)
    metrics['roe'] = roe * 100 if not pd.isna(roe) else np.nan
    
    # 5. Dividend Growth 5Y (10%)
    try:
        if len(dividends) > 0:
            # Agrupa dividendos por ano
            div_annual = dividends.resample('Y').sum()
            if len(div_annual) >= 5:
                years_ago_5 = div_annual.iloc[-5]
                current = div_annual.iloc[-1]
                if years_ago_5 > 0:
                    metrics['div_growth_5y'] = ((current / years_ago_5) ** (1/5) - 1) * 100
                else:
                    metrics['div_growth_5y'] = np.nan
            else:
                metrics['div_growth_5y'] = np.nan
        else:
            metrics['div_growth_5y'] = np.nan
    except:
        metrics['div_growth_5y'] = np.nan
    
    # 6. Earnings Growth 5Y (10%)
    earnings_growth = info.get('earningsQuarterlyGrowth', np.nan)
    metrics['earnings_growth_5y'] = earnings_growth * 100 if not pd.isna(earnings_growth) else np.nan
    
    # 7. Revenue Growth 5Y (5%)
    revenue_growth = info.get('revenueGrowth', np.nan)
    metrics['revenue_growth_5y'] = revenue_growth * 100 if not pd.isna(revenue_growth) else np.nan
    
    # 8. Dividend Track Record (5%)
    try:
        if len(dividends) > 0:
            first_div = dividends.index[0]
            years_paying = (datetime.now() - first_div).days / 365.25
            metrics['dividend_years'] = years_paying
        else:
            metrics['dividend_years'] = 0
    except:
        metrics['dividend_years'] = 0
    
    # 9. Price Strength (5%)
    try:
        if len(hist) > 0:
            ytd_return = ((hist['Close'][-1] / hist['Close'][0]) - 1) * 100
            # Comparar com S&P 500
            spy = yf.Ticker("^GSPC")
            spy_hist = spy.history(period="1y")
            if len(spy_hist) > 0:
                spy_return = ((spy_hist['Close'][-1] / spy_hist['Close'][0]) - 1) * 100
                metrics['price_strength'] = ytd_return - spy_return
            else:
                metrics['price_strength'] = ytd_return
        else:
            metrics['price_strength'] = np.nan
    except:
        metrics['price_strength'] = np.nan
    
    # 10. Beta (10%)
    beta = info.get('beta', np.nan)
    metrics['beta'] = beta if not pd.isna(beta) else np.nan
    
    return metrics

def calculate_bsd_score(metrics):
    """Calcula a pontuação BSD final (0-100)"""
    total_score = 0
    
    for criterion, weight in WEIGHTS.items():
        value = metrics.get(criterion, np.nan)
        score = calculate_score(value, criterion)
        total_score += score * weight
    
    return round(total_score, 2)
```

```python
# =============================================================================
# ANÁLISE PRINCIPAL
# =============================================================================

def analyze_stocks(tickers):
    """Analisa uma lista de tickers e retorna ranking BSD"""
    results = []
    errors = []
    
    print("Iniciando análise BSD...")
    print("=" * 80)
    
    for ticker in tickers:
        print(f"\nAnalisando {ticker}...")
        try:
            metrics = calculate_metrics(ticker)
            
            if metrics:
                bsd_score = calculate_bsd_score(metrics)
                metrics['bsd_score'] = bsd_score
                results.append(metrics)
                print(f"✓ {ticker}: Pontuação BSD = {bsd_score:.2f}")
            else:
                print(f"✗ {ticker}: Dados insuficientes")
                errors.append(ticker)
        except Exception as e:
            print(f"✗ {ticker}: Erro - {str(e)}")
            errors.append(ticker)
    
    if not results:
        print("\n⚠ Nenhuma ação foi analisada com sucesso!")
        return pd.DataFrame()
    
    # Criar DataFrame e ordenar por pontuação
    df = pd.DataFrame(results)
    df = df.sort_values('bsd_score', ascending=False).reset_index(drop=True)
    df.index = df.index + 1  # Ranking começa em 1
    
    if errors:
        print(f"\n⚠ Ações com erro na análise: {', '.join(errors)}")
    
    return df
```
