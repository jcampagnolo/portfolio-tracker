```python
import yfinance as yf
import pandas as pd
import numpy as np
```

```python
def calcular_valuation(ticker,
                       crescimento=0.05,
                       desconto=0.10,
                       crescimento_terminal=0.03,
                       anos=5,
                       margem_seguranca=0.30):
    try:
        empresa = yf.Ticker(ticker)
        
        fcf = empresa.cashflow.loc['Free Cash Flow'][0]
        shares = empresa.info['sharesOutstanding']
        preco_atual = empresa.history(period="1d")['Close'][0]
        
        fluxos = []
        fcf_atual = fcf
        
        for i in range(1, anos+1):
            fcf_atual = fcf_atual * (1 + crescimento)
            valor_presente = fcf_atual / ((1 + desconto) ** i)
            fluxos.append(valor_presente)
        
        valor_terminal = (fcf_atual * (1 + crescimento_terminal)) / (desconto - crescimento_terminal)
        valor_terminal_presente = valor_terminal / ((1 + desconto) ** anos)
        
        valor_empresa = sum(fluxos) + valor_terminal_presente
        valor_por_acao = valor_empresa / shares
        
        valor_margem = valor_por_acao * (1 - margem_seguranca)
        upside = (valor_por_acao / preco_atual - 1) * 100
        
        return {
            "Ticker": ticker,
            "Preço Atual": preco_atual,
            "Valor Intrínseco": valor_por_acao,
            "Preço com Margem": valor_margem,
            "Upside (%)": upside
        }
    
    except:
        return {
            "Ticker": ticker,
            "Preço Atual": None,
            "Valor Intrínseco": None,
            "Preço com Margem": None,
            "Upside (%)": None
        }
```

```python
tickers = [
    "VALE3.SA",
    "PETR4.SA",
    "WEGE3.SA",
    "ITUB4.SA",
    "AAPL",
    "MSFT",
    "GOOGL"
]

resultados = []

for t in tickers:
    resultados.append(calcular_valuation(t))

df = pd.DataFrame(resultados)

df
```
